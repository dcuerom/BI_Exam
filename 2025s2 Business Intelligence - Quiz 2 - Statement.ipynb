# Dataset
<br>
<b>Dataset file</b>: 251111_hospital.csv  
<br><br>
<div style="text-align: justify">
This dataset compiles detailed information about births that occurred in a Chilean hospital between 2019 and 2021, covering both the pre-pandemic and pandemic periods of COVID-19. Each row represents a birth and includes clinical, demographic, and contextual variables related to the mother, pregnancy, delivery, and newborn. It includes data such as the mother's age and nationality, marital status, prenatal care, type of delivery, newborn condition, sex, birth weight, gestational age, Apgar scores, and whether the newborn was hospitalized or received immediate breastfeeding. It also indicates whether there was a companion during the delivery and whether it was a multiple pregnancy. This dataset allows for comparative analysis between healthcare contexts (pandemic vs. pre-pandemic) and the evaluation of factors associated with perinatal outcomes.
</div>

## Fields  
- **periodo**: Month and year when the birth occurred, used to organize the data chronologically. **Possible values**: 01-2019, 02-2019, ..., 06-2021.  
<br>

- **periodo_estudio**: Classification of the period according to whether it occurred before or during the COVID-19 pandemic. **Possible values**: PANDEMICO, PREPANDEMICO.  
<br>

- **paridad**: Parity in obstetrics refers to the number of previous pregnancies that reached a viable gestational age. **Possible values**: PRIMIPARA, MULTIPARA, NO DESCRITO.  
<br>

- **edad_madre**: Age of the mother in years at the time of delivery. **Possible values**: Rango entre 13 y 45 años.  
<br>

- **nacionalidad**: Declared nationality of the mother. **Possible values**: CHILENA, HAITIANA, VENEZOLANA, BOLIVIANA, ECUATORIANA, CUBANA, PERUANA, COLOMBIANA, DOMINICANA, ARGENTINA, NO DESCRITO, EGIPCIA, BRASILEÑA, PARAGUAYA, CHINA, KENIATA, PASAPORTE EXTRANJERO NO DESCRITO.  
<br>

- **estado_civil_madre**: Marital status of the mother at the time of delivery. **Possible values**: CONVIVIENTE, SOLTERA, CASADA, UNION CIVIL, DIVORCIADA, NO DESCRITO.  
<br>

- **lugar_control_comuna**: Municipality where the mother received prenatal care. **Possible values**: RANCAGUA, SANTA CRUZ, REQUINOA, etc. (municipalities in Chile), EXTRASISTEMA, SIN CONTROL, NO DESCRITO.  
<br>

- **embarazo_controlado**: Indicates whether the pregnancy was monitored by healthcare professionals. **Possible values**: CONTROLADO, SIN CONTROL.  
<br>

- **embarazo_gemelar**: Indicates whether the pregnancy was multiple (e.g., twins) or not. **Possible values**: SI, NO.  
<br>

- **lugar_parto**: Indicates whether the delivery occurred in a hospital or elsewhere. **Possible values**: PARTO HOSP, PARTO NO HOSP.  
<br>

- **tipo_parto**: Type of delivery. **Possible values**: CESAREA, PARTO VAGINAL.  
<br>

- **tipo_cesarea**: Classification of the cesarean section. **Possible values**: ELECTIVA, URGENCIA, NO APLICA, NO DESCRITO.  
<br>

- **condicion_rn**: Outcome of the birth in terms of the newborn’s vital condition. **Possible values**: NACIDO VIVO, ABORTO, MORTINATO, FALLECE EN INMEDIATO, IVE.  
<br>

- **sexo_rn**: Sex assigned to the newborn. **Possible values**: MASCULINO, FEMENINO, INDEFERENCIADO.  
<br>

- **peso_nacimiento**: Newborn’s weight at birth in grams. **Possible values**: Rango entre 0 y 5215 gramos.  
<br>

- **edad_gestacional**: Gestational age in weeks at the time of delivery. **Possible values**: Rango entre 9 y 42 semanas.  
<br>

- **antropometria**: Classification of the newborn based on weight and gestational age. **Possible values**: PEG, AEG, GEG, PEGS, NO APLICA.  
<br>

- **apgar_1min**: Apgar score at the first minute of life (0 to 10). **Possible values**: 0 a 10.  
<br>

- **apgar_5min**: Apgar score at the fifth minute of life (0 to 10). **Possible values**: 0 a 10.  
<br>

- **hospitalizado**: Indicates whether the newborn was hospitalized or died immediately. **Possible values**: SI, NO, NO APLICA.  
<br>

- **lm_inmediato**: Indicates whether the newborn received immediate breastfeeding after birth. **Possible values**: SI, NO, NO APLICA, NO DESCRITO.  
<br>

- **acompanante**: Indicates whether there was a companion present during delivery. **Possible values**: SI, NO, NO DESCRITO.
